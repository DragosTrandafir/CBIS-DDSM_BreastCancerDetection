```python
import os

def count_folders_with_one_image(root_folder):
    # Define allowed image extensions
    image_extensions = {'.png'}
    
    count = 0  # Number of folders that contain exactly 1 image

    # Traverse all subdirectories
    for dirpath, dirnames, filenames in os.walk(root_folder):
        # Count only image files in this directory (not recursive)
        image_files = [f for f in filenames if os.path.splitext(f)[1].lower() in image_extensions]

        if len(image_files) == 1:
            count += 1
            print(f"Folder with 1 image: {dirpath} -> {image_files[0]}")

    print(f"\nTotal folders containing exactly 1 image: {count}")
    return count


folder_path = "D:/cbis-ddsm_dataset_licenta/data/processed_separate_roi_full_masks/calc_case_description_test_set_png/full_images"

if not os.path.isdir(folder_path):
    print("Error: The provided path is not a valid folder.")
else:
    count_folders_with_one_image(folder_path)
```

# See how many full images have more tumors

## See how many patients have more tumors

```python
import os

def count_patients_with_more_tumors(full_images_folder, masks_images_folder):
    # Define allowed image extensions
    image_extensions = {'.png'}
    count = 0

    # Walk through all folders in full_images_folder
    for dirpath, dirnames, filenames in os.walk(full_images_folder):
        # Get relative path (e.g., "patient_01/case_02")
        relative_path = os.path.relpath(dirpath, full_images_folder)
        
        # Build the matching path inside masks_images_folder
        masks_dirpath = os.path.join(masks_images_folder, relative_path)

        # Skip if corresponding folder doesn’t exist in masks_images_folder
        if not os.path.exists(masks_dirpath):
            print(f"⚠️ Missing corresponding folder for: {dirpath}")
            continue

        # List image files in both folders
        image_files_full = [f for f in os.listdir(dirpath)
                            if os.path.splitext(f)[1].lower() in image_extensions]
        image_files_masks = [f for f in os.listdir(masks_dirpath)
                             if os.path.splitext(f)[1].lower() in image_extensions]

        count_full = len(image_files_full)
        count_masks = len(image_files_masks)

        # Compare the two counts
        if count_full != count_masks:
            count += 1
            print(f"❗ Folder mismatch: {relative_path} -> full={count_full}, masks={count_masks}")

    print(f"\n✅ Total patients with different number of images: {count}")
    return count


# Example usage
folder_path1 = "D:/cbis-ddsm_dataset_licenta/data/processed_separate_roi_full_masks/calc_case_description_test_set_png/full_images"
folder_path2 = "D:/cbis-ddsm_dataset_licenta/data/processed_separate_roi_full_masks/calc_case_description_test_set_png/masks"

count_total = count_patients_with_more_tumors(folder_path1, folder_path2)
print(f"Total count: {count_total}")
```
